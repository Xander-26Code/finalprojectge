```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
data = np.loadtxt('Final_Project_export/data1.txt', delimiter=',')
x = data[:, 0]
y = data[:, 1]
plt.scatter(x, y, color='blue')
```

```python
def polyFunction(x, a0, a1, a2):
    return a0 + a1 * x + a2 * x ** 2

def expoFunction(x, a0, a1):
    return a0 * np.exp(a1 * x)

```

```python
popt_poly = curve_fit(polyFunction, x, y)[0]
popt_expo = curve_fit(expoFunction, x, y)[0]
print(popt_poly)
print(popt_expo)
```
